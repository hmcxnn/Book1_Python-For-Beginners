```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.stats import gaussian_kde

# 加载数据
iris_sns = sns.load_dataset("iris")

fig, ax = plt.subplots(figsize=(10, 7))

# 绘制 fill 图
sns.kdeplot(data=iris_sns, x="sepal_length", hue="species", 
            multiple="fill", bw_adjust=0.5, ax=ax)

# ===== 画 y=0.5 水平线 =====
ax.axhline(y=0.5, color='red', linestyle='--', linewidth=2, label='y=0.5')

# ===== 计算分界线 x 值 =====
# 获取各类别数据
setosa = iris_sns[iris_sns['species'] == 'setosa']['sepal_length'].values
versicolor = iris_sns[iris_sns['species'] == 'versicolor']['sepal_length'].values
virginica = iris_sns[iris_sns['species'] == 'virginica']['sepal_length'].values

# 计算 KDE
x_range = np.linspace(4, 8, 1000)
bw = 0.5

kde_setosa = gaussian_kde(setosa, bw_method=bw)(x_range)
kde_versicolor = gaussian_kde(versicolor, bw_method=bw)(x_range)
kde_virginica = gaussian_kde(virginica, bw_method=bw)(x_range)

# 归一化
total = kde_setosa + kde_versicolor + kde_virginica
prop_setosa = kde_setosa / total
cumsum_setosa = prop_setosa

# 找 setosa 占比 = 50% 的位置（第一个分界线）
idx_50 = np.argmin(np.abs(prop_setosa - 0.5))
x_50 = x_range[idx_50]

# 找 setosa + versicolor = 50% 的位置（第二个分界线）
cumsum_versi = cumsum_setosa + kde_versicolor / total
idx_cum_50 = np.argmin(np.abs(cumsum_versi - 0.5))
x_cum_50 = x_range[idx_cum_50]

# ===== 标注分界线 =====
# 标注 setosa=50% 的点
ax.axvline(x=x_50, color='blue', linestyle=':', alpha=0.6)
ax.plot(x_50, 0.5, 'bo', markersize=8)
ax.annotate(f'setosa=50%\nx={x_50:.2f}', 
            xy=(x_50, 0.5), xytext=(x_50-0.5, 0.65),
            arrowprops=dict(arrowstyle='->', color='blue'),
            fontsize=9, color='blue',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

# 标注 setosa+versicolor=50% 的点  
ax.axvline(x=x_cum_50, color='green', linestyle=':', alpha=0.6)
ax.plot(x_cum_50, 0.5, 'go', markersize=8)
ax.annotate(f'setosa+versi=50%\nx={x_cum_50:.2f}', 
            xy=(x_cum_50, 0.5), xytext=(x_cum_50+0.2, 0.35),
            arrowprops=dict(arrowstyle='->', color='green'),
            fontsize=9, color='green',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

ax.set_title('Fill KDE with 50% Boundaries', fontsize=12)
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()

print(f"分界线 x 值：")
print(f"  setosa 占比 = 50%: x = {x_50:.2f}")
print(f"  setosa + versicolor = 50%: x = {x_cum_50:.2f}")
```

### 协方差矩阵

```python
f,(ax1,ax2,ax3) = plt.subplots(1,3,sharey=True)
features = list(iris_sns.columns)[:-1]

g1 = sns.heatmap(iris_sns.loc[iris_sns.species == 'setosa', features].cov(),
                 cmap="RdYlBu_r",fmt='.2f',
                 annot=True,cbar=False,ax=ax1,square=True,
                 vmax = 0.4, vmin = 0)
ax1.set_title('Setosa')

g2 = sns.heatmap(iris_sns.loc[iris_sns.species == 'versicolor', features].cov(),
                 cmap="RdYlBu_r",fmt='.2f',
                 annot=True,cbar=False,ax=ax2,square=True,
                 vmax = 0.4, vmin = 0)
ax2.set_title('Versicolor')

g3 = sns.heatmap(iris_sns.loc[iris_sns.species == 'virginica', features].cov(),
                 cmap="RdYlBu_r",fmt='.2f',
                 annot=True,cbar=False,ax=ax3,square=True,
                 vmax = 0.4, vmin = 0)
ax3.set_title('Virginica')
```
