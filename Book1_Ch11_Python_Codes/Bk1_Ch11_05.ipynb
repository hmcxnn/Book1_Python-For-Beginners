# 用Seaborn生成热图

```python
# 导入包
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 从seaborn中导入鸢尾花样本数据
iris_sns = sns.load_dataset("iris")
```

```python
# 绘制热图
fig, ax = plt.subplots()

sns.heatmap(data=iris_sns.iloc[:,0:-1],
            vmin = 0, vmax = 8,
            ax = ax,
            yticklabels = False,
            xticklabels = ['Sepal length', 'Sepal width',
                           'Petal length', 'Petal width'],
            cmap = 'RdYlBu_r')
```
