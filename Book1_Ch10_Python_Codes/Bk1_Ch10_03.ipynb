# 用add_subplot() 方法创建一个新的子图对象
Book_1《编程不难》 | 鸢尾花书：从加减乘除到机器学习 

```python
import numpy as np
import matplotlib.pyplot as plt 
 
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)
y_1 = np.cos(x)
y_min,y_max = -1.5,1.5
y_ticks = np.arange(-1.5,2,0.5)
fig = plt.figure()
ax = fig.add_subplot(1, 2, 1)
ax.set_xlim(0,2*np.pi)
ax.set_ylim(y_min,y_max)
x_ticks = np.arange(0, 2*np.pi+np.pi/2, np.pi)
x_ticklabels = [r'$0$', 
                r'$\pi$', 
                r'$2\pi$']
ax.set_xticks(x_ticks)
ax.set_xticklabels(x_ticklabels)
ax.set_ylabel(r'$f(x)$',rotation=0)
ax.set_xlabel(r'$x$',rotation=0)
ax.set_yticks(y_ticks)
ax.set_aspect('equal') 
ax.grid(True)
ax.set_title("Sine function")
ax.plot(x, y)

ax2 = fig.add_subplot(1, 2, 2)
ax2.set_xlim(0,2*np.pi)
ax2.set_xticks(x_ticks)
ax2.set_xticklabels(x_ticklabels)
ax2.set_ylabel(r'$f(x)$',rotation=0)
ax2.set_xlabel(r'$x$',rotation=0)
ax2.set_ylim(y_min,y_max)
ax2.set_yticks(y_ticks) 
ax2.set_yticklabels([])
ax2.set_aspect('equal') 
ax2.grid(True)
ax2.set_title("Cosine function")
ax2.plot(x, y_1,color='r')

plt.tight_layout(pad=1)
plt.show()
```
